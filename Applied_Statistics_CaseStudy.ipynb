```python
# ============================================================
# E-COMMERCE RETAIL SALES ANALYSIS AND DEMAND FORECASTING
# Using Statistical Methods and Python
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy import stats
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

from sklearn.metrics import mean_absolute_error, mean_squared_error


# ============================================================
# 2. LOAD DATASET
# ============================================================

file_path = "samplesuperstore.csv"

df = pd.read_csv(file_path)

print("\n================ DATASET INFORMATION ================\n")

print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
print(df.head())


# ============================================================
# 3. DATA PREPROCESSING
# ============================================================

print("\n================ DATA PREPROCESSING ================\n")

# Convert Order Date into datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Fill missing numerical values with median
numeric_columns = ["Sales", "Quantity", "Discount", "Profit"]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("\nDataset after preprocessing:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ============================================================
# MODULE 1
# DESCRIPTIVE STATISTICS
# ============================================================

print("\n\n======================================================")
print("MODULE 1 - DESCRIPTIVE STATISTICS")
print("======================================================")

# Select important numerical variables
data = df[["Sales", "Quantity", "Discount", "Profit"]]

print("\nDescriptive Statistics:")
print(data.describe())

# Mean
print("\nMean:")
print(data.mean())

# Median
print("\nMedian:")
print(data.median())

# Mode
print("\nMode:")
print(data.mode().iloc[0])

# Variance
print("\nVariance:")
print(data.var())

# Standard deviation
print("\nStandard Deviation:")
print(data.std())

# Minimum
print("\nMinimum:")
print(data.min())

# Maximum
print("\nMaximum:")
print(data.max())


# ============================================================
# 4. CORRELATION AND COVARIANCE
# ============================================================

print("\n================ CORRELATION ================\n")

print("Covariance Matrix:")
print(data.cov())

print("\nPearson Correlation:")
print(data.corr(method="pearson"))

print("\nSpearman Correlation:")
print(data.corr(method="spearman"))


# -----------------------------
# Correlation Heatmap
# -----------------------------

plt.figure(figsize=(8, 6))

correlation = data.corr()

plt.imshow(correlation, cmap="coolwarm", interpolation="none")

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Matrix of Retail Variables")

plt.tight_layout()
plt.show()


# ============================================================
# 5. SALES VS PROFIT
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Sales"],
    df["Profit"],
    alpha=0.5
)

plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit")

plt.grid(True)
plt.show()


# ============================================================
# MODULE 1 - PROBABILITY DISTRIBUTION
# ============================================================

print("\n================ PROBABILITY ANALYSIS ================\n")

sales = df["Sales"]

# Mean and standard deviation
sales_mean = sales.mean()
sales_std = sales.std()

print("Mean Sales :", sales_mean)
print("Sales Std  :", sales_std)


# -----------------------------
# Histogram + Normal Distribution
# -----------------------------

plt.figure(figsize=(9, 5))

plt.hist(
    sales,
    bins=40,
    density=True,
    alpha=0.6
)

x = np.linspace(
    sales.min(),
    sales.max(),
    500
)

normal_pdf = stats.norm.pdf(
    x,
    sales_mean,
    sales_std
)

plt.plot(
    x,
    normal_pdf,
    linewidth=2
)

plt.xlabel("Sales")
plt.ylabel("Density")

plt.title(
    "Sales Distribution with Normal Probability Density"
)

plt.show()


# ============================================================
# 6. EMPIRICAL CDF
# ============================================================

sorted_sales = np.sort(sales)

cdf = np.arange(
    1,
    len(sorted_sales) + 1
) / len(sorted_sales)

plt.figure(figsize=(8, 5))

plt.plot(
    sorted_sales,
    cdf
)

plt.xlabel("Sales")
plt.ylabel("Cumulative Probability")

plt.title("Empirical CDF of Sales")

plt.grid(True)
plt.show()


# ============================================================
# MODULE 2
# CONTINUOUS RANDOM VARIABLES
# ============================================================

print("\n\n======================================================")
print("MODULE 2 - CONTINUOUS RANDOM VARIABLES")
print("======================================================")

# Probability of Sales being between two values

lower_limit = 100
upper_limit = 500

probability = np.mean(
    (sales >= lower_limit) &
    (sales <= upper_limit)
)

print(
    f"Probability of Sales between "
    f"{lower_limit} and {upper_limit}:",
    probability
)


# -----------------------------
# Normal Distribution Probability
# -----------------------------

z_probability = (
    stats.norm.cdf(upper_limit, sales_mean, sales_std)
    -
    stats.norm.cdf(lower_limit, sales_mean, sales_std)
)

print(
    "Normal Distribution Probability:",
    z_probability
)


# ============================================================
# MODULE 3
# HYPOTHESIS TESTING
# ============================================================

print("\n\n======================================================")
print("MODULE 3 - HYPOTHESIS TESTING")
print("======================================================")


# ============================================================
# ONE SAMPLE T-TEST
# ============================================================

# H0: Mean sales = 500
# H1: Mean sales != 500

benchmark = 500

t_statistic, p_value = stats.ttest_1samp(
    sales,
    benchmark
)

print("\nOne Sample T-Test")

print("Benchmark Mean :", benchmark)
print("T-Statistic    :", t_statistic)
print("P-Value        :", p_value)

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject Null Hypothesis")
else:
    print("Decision: Fail to Reject Null Hypothesis")


# ============================================================
# TWO SAMPLE T-TEST
# ============================================================

# Compare sales between two regions

east_sales = df[
    df["Region"] == "East"
]["Sales"]

west_sales = df[
    df["Region"] == "West"
]["Sales"]

t_statistic, p_value = stats.ttest_ind(
    east_sales,
    west_sales,
    equal_var=False
)

print("\nTwo Sample T-Test: East vs West")

print("East Mean Sales :", east_sales.mean())
print("West Mean Sales :", west_sales.mean())

print("T-Statistic :", t_statistic)
print("P-Value     :", p_value)

if p_value < 0.05:
    print("Decision: Significant difference exists")
else:
    print("Decision: No significant difference")


# ============================================================
# ANOVA TEST
# ============================================================

print("\n================ ANOVA TEST ================\n")

# Sales among different product categories

category_groups = []

for category in df["Category"].unique():

    category_sales = df[
        df["Category"] == category
    ]["Sales"]

    category_groups.append(category_sales)


f_statistic, p_value = stats.f_oneway(
    *category_groups
)

print("F-Statistic :", f_statistic)
print("P-Value     :", p_value)

if p_value < 0.05:
    print(
        "Decision: Significant difference exists "
        "between category sales."
    )
else:
    print(
        "Decision: No significant difference "
        "between category sales."
    )


# ============================================================
# MODULE 4
# TIME SERIES ANALYSIS
# ============================================================

print("\n\n======================================================")
print("MODULE 4 - TIME SERIES ANALYSIS")
print("======================================================")


# ============================================================
# MONTHLY SALES
# ============================================================

monthly_sales = (
    df
    .set_index("Order Date")
    .resample("MS")["Sales"]
    .sum()
)

print("\nMonthly Sales:")
print(monthly_sales.head(12))


# ============================================================
# MONTHLY SALES TREND
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.xlabel("Month")
plt.ylabel("Sales")

plt.title("Monthly Retail Sales Trend")

plt.grid(True)
plt.tight_layout()

plt.show()


# ============================================================
# MOVING AVERAGE
# ============================================================

# 3-month Simple Moving Average

monthly_sales_df = pd.DataFrame({
    "Sales": monthly_sales
})

monthly_sales_df["SMA_3"] = (
    monthly_sales_df["Sales"]
    .rolling(window=3)
    .mean()
)

# 3-month Exponential Moving Average

monthly_sales_df["EMA_3"] = (
    monthly_sales_df["Sales"]
    .ewm(span=3, adjust=False)
    .mean()
)

print("\nMoving Average Results:")
print(monthly_sales_df.head(10))


# ============================================================
# SMA AND EMA GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales_df.index,
    monthly_sales_df["Sales"],
    label="Actual Sales"
)

plt.plot(
    monthly_sales_df.index,
    monthly_sales_df["SMA_3"],
    label="3-Month SMA"
)

plt.plot(
    monthly_sales_df.index,
    monthly_sales_df["EMA_3"],
    label="3-Month EMA"
)

plt.xlabel("Month")
plt.ylabel("Sales")

plt.title(
    "Actual Sales vs Simple and Exponential Moving Average"
)

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# ACF
# ============================================================

plt.figure(figsize=(10, 5))

plot_acf(
    monthly_sales.dropna(),
    lags=15
)

plt.title("ACF of Monthly Sales")

plt.tight_layout()
plt.show()


# ============================================================
# PACF
# ============================================================

plt.figure(figsize=(10, 5))

plot_pacf(
    monthly_sales.dropna(),
    lags=10,
    method="ywm"
)

plt.title("PACF of Monthly Sales")

plt.tight_layout()
plt.show()


# ============================================================
# MODULE 4 - ARIMA FORECASTING
# ============================================================

print("\n\n======================================================")
print("ARIMA SALES FORECASTING")
print("======================================================")


# ARIMA model
# p = 1
# d = 1
# q = 1

model = ARIMA(
    monthly_sales,
    order=(1, 1, 1)
)

model_fit = model.fit()

print("\nARIMA Model Summary:")
print(model_fit.summary())


# ============================================================
# FORECAST NEXT 6 MONTHS
# ============================================================

forecast_steps = 6

forecast = model_fit.forecast(
    steps=forecast_steps
)

forecast_dates = pd.date_range(
    start=monthly_sales.index[-1] +
          pd.DateOffset(months=1),
    periods=forecast_steps,
    freq="MS"
)

forecast_df = pd.DataFrame({
    "Month": forecast_dates,
    "Forecasted Sales": forecast.values
})

print("\n================ SALES FORECAST ================\n")

print(forecast_df)


# ============================================================
# ACTUAL VS FORECAST GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    label="Historical Sales"
)

plt.plot(
    forecast_dates,
    forecast.values,
    marker="o",
    linewidth=2,
    label="Forecasted Sales"
)

plt.xlabel("Month")
plt.ylabel("Sales")

plt.title(
    "Retail Sales Forecast Using ARIMA"
)

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# MODULE 5
# MODEL EVALUATION
# ============================================================

print("\n\n======================================================")
print("MODULE 5 - MODEL EVALUATION")
print("======================================================")


# ============================================================
# TRAIN-TEST SPLIT
# ============================================================

train_size = int(
    len(monthly_sales) * 0.80
)

train = monthly_sales.iloc[:train_size]

test = monthly_sales.iloc[train_size:]


print("\nTraining Data:", len(train))
print("Testing Data :", len(test))


# ============================================================
# TRAIN ARIMA MODEL
# ============================================================

train_model = ARIMA(
    train,
    order=(1, 1, 1)
)

train_fit = train_model.fit()


# Forecast test period

test_forecast = train_fit.forecast(
    steps=len(test)
)


# ============================================================
# MAE AND RMSE
# ============================================================

mae = mean_absolute_error(
    test,
    test_forecast
)

rmse = np.sqrt(
    mean_squared_error(
        test,
        test_forecast
    )
)


print("\n================ MODEL PERFORMANCE ================\n")

print("MAE  :", mae)
print("RMSE :", rmse)


# ============================================================
# ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    train.index,
    train.values,
    label="Training Data"
)

plt.plot(
    test.index,
    test.values,
    label="Actual Test Sales"
)

plt.plot(
    test.index,
    test_forecast,
    marker="o",
    label="Predicted Sales"
)

plt.xlabel("Month")
plt.ylabel("Sales")

plt.title(
    "ARIMA Model - Actual vs Predicted Sales"
)

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# MODEL PERFORMANCE TABLE
# ============================================================

performance = pd.DataFrame({
    "Model": ["ARIMA(1,1,1)"],
    "MAE": [mae],
    "RMSE": [rmse]
})

print("\n================ MODEL PERFORMANCE TABLE ================\n")

print(performance)


# ============================================================
# CATEGORY-WISE SALES
# ============================================================

print("\n================ CATEGORY ANALYSIS ================\n")

category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)


plt.figure(figsize=(8, 5))

category_sales.plot(
    kind="bar"
)

plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.title("Sales by Product Category")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# REGION-WISE SALES
# ============================================================

region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nRegion-wise Sales:")
print(region_sales)


plt.figure(figsize=(8, 5))

region_sales.plot(
    kind="bar"
)

plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.title("Sales by Region")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# TOP 10 PRODUCTS
# ============================================================

top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop 10 Products:")
print(top_products)


plt.figure(figsize=(10, 6))

top_products.sort_values().plot(
    kind="barh"
)

plt.xlabel("Sales")
plt.ylabel("Product")

plt.title("Top 10 Products by Sales")

plt.tight_layout()
plt.show()


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n\n======================================================")
print("PROJECT SUMMARY")
print("======================================================")

print("\nTotal Records:", len(df))

print(
    "Total Sales: {:.2f}".format(
        df["Sales"].sum()
    )
)

print(
    "Total Profit: {:.2f}".format(
        df["Profit"].sum()
    )
)

print(
    "Average Sales: {:.2f}".format(
        df["Sales"].mean()
    )
)

print(
    "Average Profit: {:.2f}".format(
        df["Profit"].mean()
    )
)

print(
    "Average Quantity: {:.2f}".format(
        df["Quantity"].mean()
    )
)

print(
    "ARIMA MAE: {:.2f}".format(mae)
)

print(
    "ARIMA RMSE: {:.2f}".format(rmse)
)

print("\nNext 6 Months Sales Forecast:")
print(forecast_df)

print("\n======================================================")
print("PROJECT COMPLETED")
print("======================================================")
```
